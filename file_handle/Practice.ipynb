```python
file_content = []
for file in os.listdir():
    if ".txt" in file:
        with open(file,"r") as f:
            file_content.append(f.read())

```

```python
# Create new file and add all this list contain in it 
file_name = os.path.join(directory,"combineFile.txt")

with open(file_name,"w") as f:
    for line in file_content :
        f.write(line)


print("Content transfer successfully . . .")
```

```python
# move all files into new directory
import shutil

os.mkdir("rawFiles")

destination = "D:\\Other\\Python\\file_handle\\Assignment2\\rawFiles\\"

source = "D:\\Other\\Python\\file_handle\\Assignment2\\"

for i in range(11):
    file_name = os.path.join(source,f"text_{i}.txt")
    shutil.move(file_name,destination)

print("Move sucessfully . . .")

    

    
```
