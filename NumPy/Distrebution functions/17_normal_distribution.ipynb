# normal(Gaussian) distribution - very important.
# random.normal() method - loc(mean) ,scale(standered deviation), size

```python
from numpy import random
```

```python
# normal distribution of size 2*3
x = random.normal(size=(2,3))
print(x)
```

```python
# generate random normal dist of size 2*3 with mean at 1 and standard deviation of 2 :

x = random.normal(loc=1 , scale=2, size=(2,3))
print(x);
```

# Visulization of normal distribution

# the curve of a normal dist is also called as bell curve.

```python
from numpy import random
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.distplot(random.normal(size=1000), hist=False)
plt.show()
```
